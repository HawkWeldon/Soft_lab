```python
# Step 1: Read the Spreadsheet Data

import openpyxl, pprint

print('Opening workbook...')
wb = openpyxl.load_workbook('/Users/sakina/Desktop/cs_699/india_censuspopdata.xlsx')
sheet = wb['Population by Census Unit']
state_data = {}

print('Reading rows...')
for row in range(2, sheet.max_row + 1):
    state    = sheet['B' + str(row)].value
    district = sheet['C' + str(row)].value
    pop      = sheet['D' + str(row)].value

    # Make sure the key for this state exists.
    state_data.setdefault(state, {})
    # Make sure the key for this district in this state exists.
    state_data[state].setdefault(district, {'units': 0, 'pop': 0})

    # Tally up the census-unit count and population.
    state_data[state][district]['units'] += 1
    state_data[state][district]['pop'] += int(pop)

print('Writing results...')
result_file = open('india_census2011.py', 'w')
result_file.write('allData = ' + pprint.pformat(state_data))
result_file.close()
print('Done.')

print()
print('--- Preview ---')
pprint.pprint(state_data)
```

```python
# Project 10: Update a Spreadsheet

import openpyxl

wb = openpyxl.load_workbook('/Users/sakina/Desktop/cs_699/produceSales3.xlsx')
sheet = wb['Sheet']
```

```python
# The produce types and their updated prices
PRICE_UPDATES = {'Garlic': 3.07,
                 'Celery': 1.19,
                 'Lemon': 1.27}
```

```python
# Loop through the rows and update the prices.
for row_num in range(2, sheet.max_row + 1):  # Skip the first row.
    produce_name = sheet.cell(row=row_num, column=1).value
    if produce_name in PRICE_UPDATES:
        sheet.cell(row=row_num, column=2).value = PRICE_UPDATES[produce_name]

wb.save('updatedProduceSales3.xlsx')
```

```python
# Setting the Font Style of Cells

import openpyxl
from openpyxl.styles import Font
wb = openpyxl.Workbook()      # creates a brand NEW blank workbook, in memory
sheet = wb['Sheet']            # gets its default sheet, named 'Sheet'

italic_24_font = Font(size=24, italic=True)   # create a Font style object
sheet['A1'].font = italic_24_font              # apply that style to cell A1 (still empty!)

sheet['A1'] = 'Hello, world!'   # NOW put text into A1 — it inherits the italic/24pt style already set
wb.save('styles3_example1.xlsx')          # write everything to disk as styles3.xlsx
```

```python
import openpyxl
from openpyxl.styles import Font
wb = openpyxl.Workbook()       # creates ANOTHER new blank workbook (unrelated to the first one)
sheet = wb['Sheet']
sheet.title = 'Style2'
bold_font = Font(name='Times New Roman', bold=True)
sheet['A1'].font = bold_font
sheet['A1'] = 'Bold Times New Roman'

italic_font = Font(size=24, italic=True)
sheet['B3'].font = italic_font
sheet['B3'] = '24 pt Italic'

wb.save('styles3_example2.xlsx')
```

```python
# Adjusting Rows and Columns

import openpyxl
wb = openpyxl.Workbook()
sheet = wb['Sheet']
sheet['A1'] = 'Tall row'
sheet['B2'] = 'Wide column'
sheet.row_dimensions[1].height = 70
sheet.column_dimensions['B'].width = 20
wb.save('dimensions3.xlsx')
```

```python
# Merging and Unmerging Cells

import openpyxl
wb = openpyxl.Workbook()
sheet = wb['Sheet']
sheet.merge_cells('A1:D3')  # Merge all these cells.
sheet['A1'] = 'Twelve cells merged together.'
sheet.merge_cells('C5:D5')  # Merge these two cells.
sheet['C5'] = 'Two merged cells.'
wb.save('merged3.xlsx')
```

```python
# unmerge cells

import openpyxl
wb = openpyxl.load_workbook('merged3.xlsx')
sheet = wb['Sheet']
sheet.unmerge_cells('A1:D3')  # Split these cells up.
sheet.unmerge_cells('C5:D5')
wb.save('unmerged3.xlsx')
```

```python
# Freezing Panes

import openpyxl
wb = openpyxl.load_workbook('produceSales3.xlsx')
sheet = wb.active
sheet.freeze_panes = 'A2'  # Freeze the rows above A2.
wb.save('freezeExample3.xlsx')
```

```python
# Charts

import openpyxl
wb = openpyxl.Workbook()
sheet = wb.active
for i in range(1, 11):  # Create some data in column A.
    sheet['A' + str(i)] = i * i

ref_obj = openpyxl.chart.Reference(sheet, 1, 1, 1, 10)

series_obj = openpyxl.chart.Series(ref_obj, title='First series')

chart_obj = openpyxl.chart.BarChart()
chart_obj.title = 'My Chart'
chart_obj.append(series_obj)

sheet.add_chart(chart_obj, 'C5')
wb.save('sampleChart3.xlsx')
```
