```python
# 1. Opening a workbook
# Load an existing `.xlsx` file into a `Workbook` object.

import openpyxl                                                  # import the openpyxl library
wb = openpyxl.load_workbook('./data.xlsx')                       # open the .xlsx file and load it into memory
type(wb)                                                         # check the object's type -> Workbook
```

An Excel spreadsheet document is called a workbook. A single workbook is saved in a file with the .xlsx extension.

Each workbook can contain multiple sheets (also called worksheets). The sheet the user is currently viewing (or last viewed before closing Excel) is called the active sheet

import openpyxl — brings the library into your script so its functions/classes become available


openpyxl.load_workbook(path) — opens an existing .xlsx file from disk and loads its entire contents into memory as a Workbook object. The argument is the file path — here, an absolute path pointing to a specific location on disk


wb — this variable now holds the in-memory representation of the whole Excel file: all its sheets, cells, and values


type(wb) — confirms what kind of object was returned. This will output:
 <class 'openpyxl.workbook.workbook.Workbook'>



```python
# 2. Getting sheets from a workbook
#create 3 sheets in the workbook 

wb.sheetnames          # list of all sheet name strings
sheet = wb['Sheet3']   # get a Worksheet object by name (dict-like access)
sheet

wb.active               # get the currently active/open sheet
sheet.title             # sheet's name as a string
```

A workbook is a container for multiple sheets. You can list all sheet names (sheetnames), fetch a specific one by name (wb['name']), or grab whichever one is currently active (wb.active)

wb.sheetnames — returns a Python list of every sheet's name in the workbook, e.g. ['Sheet1', 'Sheet2', 'Sheet3']  Useful for checking what sheets exist before trying to access one


wb['Sheet3'] — retrieves a specific sheet by its exact name, using square-bracket syntax just like accessing a dictionary key	 This returns a Worksheet object, not just the name 


sheet — displaying this variable shows something like <Worksheet "Sheet3">, confirming it's a Worksheet object tied to that particular sheet inside wb


wb.active — instead of specifying a name, this grabs whichever sheet was the "active" one when the workbook was last saved 
sheet.title — once you have a Worksheet object, .title gives back its name as a plain string (e.g. 'Sheet3') : This is the reverse operation of wb['Sheet3'] instead of name → sheet, it's sheet → name


```python
# 3. Getting cells from a sheet

import openpyxl                                # import the openpyxl library
wb = openpyxl.load_workbook('data.xlsx')        # load the workbook from disk

sheet = wb['Sheet1']    # get the 'Sheet1' worksheet from the workbook
sheet['A3'].value       # get the cell at A3, then read its stored value
```

```python
c = sheet['B1']  # Get another cell from the sheet.
c.value
```

```python
f'Row {c.row}, Column {c.column} is {c.value}'
```

```python
f'Cell {c.coordinate} is {c.value}'
```

```python
sheet['C1'].value
```

c = sheet['B1'] — retrieves the Cell object at position B1 and stores it in the variable c, so it can be reused across multiple lines without re-fetching it each time


c.value — pulls out the actual data stored in that cell (here, the string 'B'). This attribute gives the real content, as opposed to just a reference to the cell object


c.row — the cell's row number as an integer
 
c.column — the cell's column number as an integer 


c.coordinate — the cell's position written the way Excel stores, as a string ('B1')
 
f-strings (f'...') — a Python feature for building readable strings by embedding variables/attributes directly inside {} placeholder


```python
# 4. Using `cell()` with row/column integers - iterating over the sheet

sheet.cell(row=1, column=2)              # get the Cell object at row 1, column 2 (i.e. B1)

sheet.cell(row=1, column=2).value        # get the value stored in that cell

for i in range(1, sheet.max_row + 1):    # loop through every row number, from 1 to the last row
    print(i, sheet.cell(row=i, column=2).value)  # print the row number and the value in column 2 (B) for that row
```

sheet.cell(row=1, column=2) — an alternative to sheet['B1'], using numeric row and column arguments instead of a coordinate string. Returns a Cell object referencing row 1, column 2 (cell B1)


sheet.cell(row=1, column=2).value — chains .value onto the result to retrieve the actual data stored in that cell, rather than just a reference to it


for i in range(1, sheet.max_row + 1): — loops through every row number in the sheet, from row 1 up to and including the last row:
1 → start at the first row


sheet.max_row + 1 → stop right after the sheet's actual last row (+1 is needed because range()'s stop value is excluded, and sheet.max_row dynamically reflects however many rows currently exist)


No step argument → defaults to 1, so every row is visited, none skipped

```python
# 6. Converting Between Column Letters and Numbers

import openpyxl                                                       # import the openpyxl library
from openpyxl.utils import get_column_letter, column_index_from_string  # import the two conversion helper functions
get_column_letter(1)   # convert column number 1 into its letter name -> 'A'
```

```python
get_column_letter(2)
```

```python
get_column_letter(27)
```

```python
get_column_letter(900)
```

```python
wb = openpyxl.load_workbook('data.xlsx')
sheet = wb['Sheet1']
get_column_letter(sheet.max_column)
```

```python
column_index_from_string('A') # Get A's number.
```

```python
column_index_from_string('AA')
```

get_column_letter(1) → 'A' — converts a column number into its Excel letter name. Column numbering starts at 1 (not 0), so 1 maps to 'A'
get_column_letter(900) → 'AHP' — demonstrates why this function is useful: past column 26 ('Z'), Excel starts using two-letter (then three-letter) column names, which get tricky to calculate by hand. This function handles that logic automatically, no matter how large the column number is
get_column_letter(sheet.max_column) — a practical combination: instead of a hardcoded number, this converts the actual last column of a real sheet into its letter name — useful when you want to know, say, "what's the rightmost column letter in this sheet?" without counting manually
column_index_from_string('AA') → 27 — the reverse operation: converts a column letter name back into its numeric position. Confirms that column 'AA' is the 27th column (after A–Z, which is 26 columns, 'AA' becomes the 27th)



```python
# 7.Getting Rows and Columns

import openpyxl                                # import the openpyxl library
wb = openpyxl.load_workbook('data.xlsx')        # load the workbook from disk
sheet = wb['Sheet1']                            # get the 'Sheet1' worksheet

sheet['A1':'C3']                                # slice: get all Cell objects in the rectangle A1 to C3
```

```python
for row_of_cell_objects in sheet['A1':'C3']:    # loop over each row in the sliced range (outer loop)
    for cell_obj in row_of_cell_objects:        # loop over each individual cell in that row (inner loop)
        print(cell_obj.coordinate, cell_obj.value)  # print the cell's position (e.g. 'A1') and its value
    print('--- END OF ROW ---')                 # after finishing one row, print a separator line
```
